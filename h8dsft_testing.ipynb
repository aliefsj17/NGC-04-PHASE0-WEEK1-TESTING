```python
# Custom Exceptions for Exception Handling
class BookAlreadyExistsError(Exception):
    """Raised when trying to add a book with an ID that already exists."""
    pass

class BookNotFoundError(Exception):
    """Raised when a book is not found in the catalog."""
    pass

class InvalidBookInputError(Exception):
    """Raised when book attributes are invalid or empty."""
    pass


class Book:
    """Represents an individual book in the library."""
    def __init__(self, book_id: str, title: str, author: str):
        if not book_id or not str(book_id).strip():
            raise InvalidBookInputError("Book ID cannot be empty.")
        if not title or not str(title).strip():
            raise InvalidBookInputError("Book title cannot be empty.")
        if not author or not str(author).strip():
            raise InvalidBookInputError("Book author cannot be empty.")
            
        self.book_id = str(book_id).strip()
        self.title = str(title).strip()
        self.author = str(author).strip()

    def __str__(self):
        return f"ID: {self.book_id} | Title: '{self.title}' | Author: {self.author}"

class LibraryCatalog:
    """Manages the collection of books with OOP principles and exception handling."""
    def __init__(self):
        # Using a dictionary with book_id as key for efficient management
        self.catalog = {}

    def add_book(self, book: Book):
        """Adds a book to the catalog with duplicate and type checking."""
        try:
            if not isinstance(book, Book):
                raise TypeError("Only instances of the Book class can be added.")
            
            if book.book_id in self.catalog:
                raise BookAlreadyExistsError(f"A book with ID '{book.book_id}' already exists.")
            
            self.catalog[book.book_id] = book
            print(f"[SUCCESS] Added: '{book.title}' by {book.author}")
            
        except (BookAlreadyExistsError, TypeError, InvalidBookInputError) as e:
            print(f"[ERROR - Add Book Failed]: {e}")
            # Re-raising or handling depending on use case; here we catch and log gracefully

    def remove_book(self, book_id: str):
        """Removes a book from the catalog using its unique identifier."""
        try:
            book_id = str(book_id).strip()
            if book_id not in self.catalog:
                raise BookNotFoundError(f"Book with ID '{book_id}' not found in the catalog.")
            
            removed_book = self.catalog.pop(book_id)
            print(f"[SUCCESS] Removed: '{removed_book.title}' (ID: {book_id})")
            return removed_book
            
        except BookNotFoundError as e:
            print(f"[ERROR - Remove Book Failed]: {e}")

    def search_books(self, query: str):
        """Searches for books by title, author, or book ID (case-insensitive, partial match)."""
        try:
            if not query or not str(query).strip():
                raise ValueError("Search query cannot be empty.")
            
            query = str(query).strip().lower()
            results = []
            
            for book in self.catalog.values():
                if query in book.title.lower() or query in book.author.lower() or query in book.book_id.lower():
                    results.append(book)
            
            if not results:
                print(f"[INFO] No books found matching query: '{query}'")
            else:
                print(f"[FOUND] {len(results)} matching book(s):")
                for book in results:
                    print(f"  -> {book}")
            return results
            
        except ValueError as e:
            print(f"[ERROR - Search Failed]: {e}")
            return []

    def display_books(self):
        """Displays all books currently in the library catalog."""
        if not self.catalog:
            print("\n📚 The library catalog is currently empty.\n")
            return
        
        print("\n================ LIBRARY CATALOG ================")
        for book in self.catalog.values():
            print(f"  {book}")
        print("=================================================\n")

# 1. Initialize the library catalog
my_library = LibraryCatalog()

print("--- 1. ADDING BOOKS ---")
# Normal additions
book1 = Book("B001", "Python Crash Course", "Eric Matthes")
book2 = Book("B002", "Clean Code", "Robert C. Martin")
book3 = Book("B003", "Fluent Python", "Luciano Ramalho")

my_library.add_book(book1)
my_library.add_book(book2)
my_library.add_book(book3)

# Test Exception: Adding duplicate book ID
print("\n--- Testing Exception: Duplicate Book ID ---")
duplicate_book = Book("B001", "Another Python Book", "Unknown")
my_library.add_book(duplicate_book)

# Test Exception: Invalid/Empty input
print("\n--- Testing Exception: Invalid Book Input ---")
try:
    invalid_book = Book("", "Book Title", "Author Name")
except InvalidBookInputError as e:
    print(f"[CAUGHT EXCEPTION]: {e}")

# Display all books
my_library.display_books()

print("--- 2. SEARCHING FOR BOOKS ---")
print("Searching by Title ('Python'):")
my_library.search_books("Python")

print("\nSearching by Author ('Martin'):")
my_library.search_books("Martin")

print("\nSearching non-existent book:")
my_library.search_books("Machine Learning")

print("\n--- 3. REMOVING BOOKS ---")
# Successful removal
my_library.remove_book("B002")

# Test Exception: Removing non-existent book ID
print("\n--- Testing Exception: Removing Non-Existent Book ---")
my_library.remove_book("B999")

# Final catalog state
my_library.display_books()
```

```python
import unittest

class TestLibraryCatalogSystem(unittest.TestCase):
    
    def setUp(self):
        """Initialize a fresh library catalog and sample books before each test method."""
        self.library = LibraryCatalog()
        self.book1 = Book("B001", "Python Crash Course", "Eric Matthes")
        self.book2 = Book("B002", "Clean Code", "Robert C. Martin")

    def test_book_creation_success(self):
        """Test valid Book object creation."""
        self.assertEqual(self.book1.book_id, "B001")
        self.assertEqual(self.book1.title, "Python Crash Course")
        self.assertEqual(self.book1.author, "Eric Matthes")

    def test_book_creation_invalid_input(self):
        """Test that invalid or empty inputs raise InvalidBookInputError."""
        with self.assertRaises(InvalidBookInputError):
            Book("", "Python 101", "Author")
        with self.assertRaises(InvalidBookInputError):
            Book("B003", "", "Author")
        with self.assertRaises(InvalidBookInputError):
            Book("B003", "Title", "   ")

    def test_add_book_success(self):
        """Test adding a book successfully to the catalog."""
        self.library.add_book(self.book1)
        self.assertIn("B001", self.library.catalog)
        self.assertEqual(self.library.catalog["B001"].title, "Python Crash Course")

    def test_add_duplicate_book_raises_error(self):
        """Test adding a book with an existing ID raises BookAlreadyExistsError."""
        self.library.add_book(self.book1)
        duplicate = Book("B001", "Different Book", "Different Author")
        with self.assertRaises(BookAlreadyExistsError):
            self.library.add_book(duplicate)

    def test_add_invalid_type_raises_error(self):
        """Test adding a non-Book object raises TypeError."""
        with self.assertRaises(TypeError):
            self.library.add_book("Not a Book object")

    def test_remove_book_success(self):
        """Test successful removal of a book from the catalog."""
        self.library.add_book(self.book1)
        removed = self.library.remove_book("B001")
        self.assertEqual(removed.title, "Python Crash Course")
        self.assertNotIn("B001", self.library.catalog)

    def test_remove_nonexistent_book_raises_error(self):
        """Test removing a non-existent book ID raises BookNotFoundError."""
        with self.assertRaises(BookNotFoundError):
            self.library.remove_book("B999")

    def test_search_books_by_title_author_id(self):
        """Test searching books by title, author, or ID keywords."""
        self.library.add_book(self.book1)
        self.library.add_book(self.book2)
        
        # Search by title keyword
        results_title = self.library.search_books("python")
        self.assertEqual(len(results_title), 1)
        self.assertEqual(results_title[0].book_id, "B001")
        
        # Search by author keyword
        results_author = self.library.search_books("martin")
        self.assertEqual(len(results_author), 1)
        self.assertEqual(results_author[0].book_id, "B002")
        
        # Search by ID
        results_id = self.library.search_books("b001")
        self.assertEqual(len(results_id), 1)
        self.assertEqual(results_id[0].title, "Python Crash Course")

    def test_search_empty_query_raises_error(self):
        """Test searching with an empty query raises ValueError."""
        with self.assertRaises(ValueError):
            self.library.search_books("")


# Run unit tests inside Jupyter Notebook without crashing the kernel
if __name__ == '__main__':
    unittest.main(argv=[''], exit=False)
```
